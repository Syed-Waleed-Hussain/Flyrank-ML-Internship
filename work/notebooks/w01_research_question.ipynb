## 3. Quick look at the data (2-3 real numbers)

To prove this lane is viable, I am loading the starter dataset below to check how many pages are actually showing a downward trend. A significant portion of declining pages means there is enough data for a machine learning model to find patterns and rank them effectively.

```python
import pandas as pd

url = 'https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv'

df = pd.read_csv(url)

total_pages = len(df)
declining_pages = len(df[df['trend_direction'] == 'down'])

print(f"Total pages in starter dataset: {total_pages:,}")
print(f"Pages currently showing a downward trend: {declining_pages:,}")
print(f"Percentage of declining pages: {(declining_pages/total_pages)*100:.1f}%")
```
