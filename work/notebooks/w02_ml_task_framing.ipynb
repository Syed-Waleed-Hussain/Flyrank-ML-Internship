## 1. My lane as an ML task (type)

For Lane 2 (Refresh / Content Opportunity Scoring), the ML task type is Supervised Learning (Scoring / Ranking). We are not just classifying 'yes' or 'no'; we need to output a probability score to rank the pages from the most urgent to the least urgent so the content team knows exactly where to start.

## 2. Target or proxy

Target: Which pages genuinely need and will benefit from a content refresh.

Proxy: Since we cannot directly measure a subjective "needs refresh" state, we use a proxy label from the historical data: is_declining (where trend_direction == 'down').

## 3. Success metric

The success metric is Precision@50 (or Precision@K). We use this because our content team has limited capacity (e.g., they can only refresh 50 pages a month). We care about how many of our top 50 recommendations are actually correct, rather than the overall accuracy across thousands of pages they will never have time to edit.

## 4. The unit of analysis, as a real dataframe

The unit of analysis is a single web page. Therefore, 1 row = 1 page (represented by content_id). Below is the dataframe proving this structure, showing the features and our target proxy column.

```python
import pandas as pd

url = 'https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv'
df = pd.read_csv(url)

unit_of_analysis_df = df[['content_id', 'impressions_90d', 'content_age_days', 'trend_direction']].copy()
# Creating our proxy target column
unit_of_analysis_df['target_proxy (is_declining)'] = unit_of_analysis_df['trend_direction'] == 'down'

print(f"Total rows in slice: {len(unit_of_analysis_df):,}")
print("Unit of analysis: 1 row = 1 content_id")
display(unit_of_analysis_df.head())
```

## 5. Why ML beats a fixed rule here

A fixed handwritten rule (like trend_direction == down AND impressions_90d > 500) is too rigid. As seen in the starter pipeline, a simple rule might only achieve a Precision@50 of 24%. An ML model (like a Random Forest) evaluates complex, overlapping patterns across multiple features (age, position tier, CTR, visibility score) simultaneously, boosting Precision@50 to around 74%. ML scales the nuance that a hardcoded rule completely misses.
